```python
import pandas as pd, numpy as np, os
fname = "dispatch_enriched.csv"
assert os.path.exists(fname), f"{fname} not found. Put it in the notebook working directory."
df = pd.read_csv(fname)
print("Loaded:", fname, "shape:", df.shape)
print("\nColumns:\n", df.columns.tolist())
print("\nNull counts (top 20):\n", df.isnull().sum().sort_values(ascending=False).head(20))
display(df.head(10))
```

```python
num_df = df.select_dtypes(include=[np.number])
print("Numeric columns:", num_df.columns.tolist())
display(num_df.describe().T)
```

```python
import matplotlib.pyplot as plt
numeric_cols = ['distance_km','booking_value','skill_match_score','driver_avg_rating','customer_repeat_rate','traffic_index','temp']
numeric_cols = [c for c in numeric_cols if c in df.columns]

for col in numeric_cols:
    plt.figure(figsize=(6,4))
    plt.hist(df[col].dropna(), bins=50)
    plt.title(f"Histogram — {col}")
    plt.xlabel(col); plt.ylabel("Count")
    plt.show()
```

```python
df['distance_km'] = np.log1p(df['distance_km'])
```

```python
df['booking_value'] = np.log1p(df['booking_value'])
```

```python
df['temp'] = df['temp'].clip(20, 40)
```

```python
import matplotlib.pyplot as plt
numeric_cols = ['distance_km','booking_value','skill_match_score','driver_avg_rating','customer_repeat_rate','traffic_index','temp']
numeric_cols = [c for c in numeric_cols if c in df.columns]

for col in numeric_cols:
    plt.figure(figsize=(6,4))
    plt.hist(df[col].dropna(), bins=50)
    plt.title(f"Histogram — {col}")
    plt.xlabel(col); plt.ylabel("Count")
    plt.show()
```

```python
corr = num_df.corr()
plt.figure(figsize=(10,8))
plt.imshow(corr, interpolation='nearest', aspect='auto')
plt.colorbar()
plt.xticks(range(len(corr.columns)), corr.columns, rotation=90)
plt.yticks(range(len(corr.index)), corr.index)
plt.title("Correlation matrix (numeric)")
plt.tight_layout()
plt.show()

# Also print top absolute correlations with target if you have 'best_driver_id' or 'success_flag'
if 'success_flag' in df.columns:
    target_corr = corr['success_flag'].abs().sort_values(ascending=False)
    print("\nTop correlations with success_flag:\n", target_corr.head(10))
```

```python
# Skill match vs distance
if 'skill_match_score' in df.columns and 'distance_km' in df.columns:
    plt.figure(figsize=(6,4))
    plt.scatter(df['distance_km'], df['skill_match_score'], s=6, alpha=0.4)
    plt.xlabel("distance_km"); plt.ylabel("skill_match_score")
    plt.title("Skill match score vs distance_km")
    plt.show()

# Skill match vs booking_value
if 'booking_value' in df.columns:
    plt.figure(figsize=(6,4))
    plt.scatter(df['booking_value'], df['skill_match_score'], s=6, alpha=0.4)
    plt.xlabel("booking_value"); plt.ylabel("skill_match_score")
    plt.title("Skill match vs booking_value")
    plt.show()
```

```python
if 'service_category' in df.columns:
    top = df['service_category'].value_counts().nlargest(15)
    plt.figure(figsize=(8,4))
    plt.bar(top.index.astype(str), top.values)
    plt.xticks(rotation=45, ha='right')
    plt.title("Top service categories (counts)")
    plt.tight_layout()
    plt.show()
    display(top)
```

```python
if 'booking_value' in df.columns:
    plt.figure(figsize=(6,3))
    plt.boxplot(df['booking_value'].dropna(), vert=False)
    plt.title("Boxplot: booking_value")
    plt.show()
    # Show top 1% values for diagnosis
    threshold = df['booking_value'].quantile(0.99)
    print("99th percentile:", threshold)
    display(df[df['booking_value'] > threshold].head(10))
```

```python
# Example: skill_match_score distribution per top 5 categories
if 'service_category' in df.columns and 'skill_match_score' in df.columns:
    top5 = df['service_category'].value_counts().nlargest(5).index.tolist()
    for cat in top5:
        subset = df[df['service_category']==cat]
        plt.figure(figsize=(6,3))
        plt.hist(subset['skill_match_score'].dropna(), bins=30)
        plt.title(f"Skill match distribution — {cat} (n={len(subset)})")
        plt.show()
```

```python
# Nulls, basic stats, top categories -> CSV for record
diagnostics = {
    "null_counts": df.isnull().sum(),
    "numeric_describe": df.select_dtypes(include=[np.number]).describe().T
}
df.isnull().sum().to_csv("diagnostics_null_counts.csv")
df.select_dtypes(include=[np.number]).describe().T.to_csv("diagnostics_numeric_describe.csv")
print("Saved diagnostics CSVs: diagnostics_null_counts.csv, diagnostics_numeric_describe.csv")
```
